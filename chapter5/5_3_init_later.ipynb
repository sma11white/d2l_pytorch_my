## 延后初始化

```python
import torch
from torch import nn

net = nn.Sequential(
    nn.LazyLinear(256),      # 只指定输出是 8,输入维度未知
    nn.ReLU(),
    nn.LazyLinear(10)
)
```

```python
print(net[0].weight)
```

```python
X = torch.rand(2, 20)
net(X)

def get_weights(m):
    if type(m) == nn.Linear:
        print(m.weight.shape, m.bias.shape)
    return None

net.apply(get_weights);
```
