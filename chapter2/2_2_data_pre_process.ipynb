## 读取数据集

```python
import os

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')  # 列名
    f.write('NA,Pave,127500\n')  # 每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')
```

```python
import pandas as pd
import torch

data_df = pd.read_csv(data_file)
data_df
```

```python
# pandas太新对于字符串处理改了,新版用这个即可
inputs, outputs = data_df.iloc[:, 0:2], data_df.iloc[:, 2]
inputs = inputs.fillna(inputs.mean(numeric_only=True))
inputs
```

```python
inputs = pd.get_dummies(inputs, dummy_na=True)
print(inputs)
```

## 练习

```python
# 1.练习1
import numpy as np

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny_test.csv')
rng = np.random.default_rng(42)
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')
    for _ in range(100):
        num_rooms = 'NA' if rng.random() < 0.15 else str(rng.integers(1, 6))
        alley = rng.choice(['Pave', 'Gravel', 'NA'], p=[0.4, 0.3, 0.3])
        price = 'NA' if rng.random() < 0.10 else str(rng.integers(50, 300) * 1000)  # 新增
        f.write(f'{num_rooms},{alley},{price}\n')
```

```python
data_df_test = pd.read_csv(data_file)
data_df_test.info()
```

```python
col_most_missing = data_df_test.isnull().sum().idxmax()
data_df_test = data_df_test.drop(columns=col_most_missing)
data_df_test.info()
```
