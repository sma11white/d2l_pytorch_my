```python
import os
import re
import hashlib
import requests

# 数据集地址（与 d2l.DATA_HUB 中的一致）
DATA_URL = 'http://d2l-data.s3-accelerate.amazonaws.com/timemachine.txt'
DATA_SHA1 = '090b5e7e70c295757f55df93cb0a180b9691891a'
DATA_DIR = '../data'
DATA_PATH = os.path.join(DATA_DIR, 'timemachine.txt')

def download_time_machine():  #@save
    """用 requests 下载时间机器数据集，并做 SHA-1 校验"""
    os.makedirs(DATA_DIR, exist_ok=True)

    # 已存在且校验通过则跳过下载
    if os.path.exists(DATA_PATH):
        with open(DATA_PATH, 'rb') as f:
            if hashlib.sha1(f.read()).hexdigest() == DATA_SHA1:
                print('文件已存在且校验通过，跳过下载')
                return DATA_PATH

    print(f'正在从 {DATA_URL} 下载...')
    response = requests.get(DATA_URL, stream=True)
    response.raise_for_status()          # 下载失败时抛出异常

    with open(DATA_PATH, 'wb') as f:
        for chunk in response.iter_content(chunk_size=8192):
            f.write(chunk)

    # 下载后校验文件完整性（可选但推荐）
    with open(DATA_PATH, 'rb') as f:
        sha1 = hashlib.sha1(f.read()).hexdigest()
    assert sha1 == DATA_SHA1, f'文件校验失败: {sha1}'
    print(f'下载完成并校验通过: {DATA_PATH}')
    return DATA_PATH

def read_time_machine():  #@save
    """将时间机器数据集加载到文本行的列表中"""
    with open(download_time_machine(), 'r') as f:
        lines = f.readlines()
    return [re.sub('[^A-Za-z]+', ' ', line).strip().lower() for line in lines]

lines = read_time_machine()
print(f'# 文本总行数: {len(lines)}')
print(lines[0])
print(lines[10])
```
